# Battery Degradation Using Partial Derivatives and the Chain Rule

## Objective
To apply the concepts of partial derivatives and the chain rule to a battery modeling scenario and understand how multivariable functions behave.

---

## Background Theory
- **Partial Derivatives** describe how a multivariable function changes as only one of its input variables changes, keeping the others constant.
- **The Chain Rule** allows us to differentiate composite functions where variables depend on one another indirectly.

---

## Activity Instructions

### 1. Battery Degradation Function
Consider the degradation function:

**D(T, C) = 0.01 * T² + 0.05 * C**

Where:
- **T** = Temperature  
- **C** = Cycle count  
- This models how battery degradation depends on both temperature and the number of charging cycles.

---

### 2. Compute Partial Derivatives

- **∂D/∂T = d/dT [0.01 * T² + 0.05 * C] = 0.02 * T**
- **∂D/∂C = d/dC [0.01 * T² + 0.05 * C] = 0.05**

---

### 3. Time Dependency
Assume that temperature and cycle count change over time as:

- **T(t) = 2t**
- **C(t) = 5t**

These describe linear increases in temperature and cycle count with respect to time.

---

### 4. Apply the Chain Rule

Use the multivariable chain rule to find the total derivative **dD/dt**:

\[
\frac{dD}{dt} = \frac{∂D}{∂T} \cdot \frac{dT}{dt} + \frac{∂D}{∂C} \cdot \frac{dC}{dt}
\]

Substituting:
- ∂D/∂T = 0.02 * T = 0.02 * (2t) = 0.04t  
- dT/dt = 2  
- ∂D/∂C = 0.05  
- dC/dt = 5  

So:

\[
\frac{dD}{dt} = (0.04t) \cdot 2 + 0.05 \cdot 5 = 0.08t + 0.25
\]

---

### 5. Evaluate at t = 3

\[
\frac{dD}{dt}\bigg|_{t=3} = 0.08 * 3 + 0.25 = 0.24 + 0.25 = \boxed{0.49}
\]

At time **t = 3**, the battery is degrading at a rate of **0.49 units per unit time**.



```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import sympy as sp

# Define symbols for symbolic computation
T, C, t = sp.symbols('T C t')

# Step 1: Define the battery degradation function
# D(T, C) = 0.01*T^2 + 0.05*C
D = 0.01*T**2 + 0.05*C

print("Step 1: Battery Degradation Function")
print("D(T, C) =", D)
print()

# Step 2: Compute partial derivatives
# ∂D/∂T - partial derivative with respect to temperature
dD_dT = sp.diff(D, T)
print("Step 2: Partial Derivatives")
print("∂D/∂T =", dD_dT)
print("This tells us how degradation changes with temperature when cycle count is held constant")
print()

# ∂D/∂C - partial derivative with respect to cycle count
dD_dC = sp.diff(D, C)
print("∂D/∂C =", dD_dC)
print("This tells us how degradation changes with cycle count when temperature is held constant")
print()

# Step 3: Define T(t) and C(t)
# T(t) = 2t and C(t) = 5t
T_t = 2*t
C_t = 5*t

print("Step 3: Temperature and Cycle Count as Functions of Time")
print("T(t) =", T_t)
print("C(t) =", C_t)
print("Both temperature and cycle count increase linearly with time")
print()

# Step 4: Use chain rule to find dD/dt
# dD/dt = (∂D/∂T)(dT/dt) + (∂D/∂C)(dC/dt)
dT_dt = sp.diff(T_t, t)
dC_dt = sp.diff(C_t, t)

print("Step 4: Chain Rule Application")
print("dT/dt =", dT_dt)
print("dC/dt =", dC_dt)
print()

# Substitute T(t) and C(t) into partial derivatives
dD_dT_sub = dD_dT.subs([(T, T_t), (C, C_t)])
dD_dC_sub = dD_dC.subs([(T, T_t), (C, C_t)])

print("After substituting T(t) and C(t):")
print("∂D/∂T =", dD_dT_sub)
print("∂D/∂C =", dD_dC_sub)
print()

# Calculate dD/dt using chain rule
dD_dt = dD_dT_sub * dT_dt + dD_dC_sub * dC_dt
dD_dt_simplified = sp.simplify(dD_dt)

print("Using the chain rule:")
print("dD/dt = (∂D/∂T)(dT/dt) + (∂D/∂C)(dC/dt)")
print("dD/dt =", dD_dt_simplified)
print("This is the total rate of degradation with respect to time")
print()

# Step 5: Evaluate dD/dt at t = 3
t_value = 3
dD_dt_at_3 = dD_dt_simplified.subs(t, t_value)

print("Step 5: Evaluate at t = 3")
print("At t =", t_value)
print("T(3) =", T_t.subs(t, t_value))
print("C(3) =", C_t.subs(t, t_value))
print("dD/dt|_{t=3} =", dD_dt_at_3)
print("This means the battery is degrading at a rate of", float(dD_dt_at_3), "units per time unit at t = 3")
print()

# Create visualizations
# 1. 3D surface plot of D(T, C)
fig = plt.figure(figsize=(15, 10))

# Create meshgrid for surface plot
T_range = np.linspace(0, 10, 50)
C_range = np.linspace(0, 20, 50)
T_mesh, C_mesh = np.meshgrid(T_range, C_range)
D_mesh = 0.01*T_mesh**2 + 0.05*C_mesh

ax1 = fig.add_subplot(221, projection='3d')
surf = ax1.plot_surface(T_mesh, C_mesh, D_mesh, cmap='viridis', alpha=0.8)
ax1.set_xlabel('Temperature (T)')
ax1.set_ylabel('Cycle Count (C)')
ax1.set_zlabel('Degradation D(T,C)')
ax1.set_title('Battery Degradation Surface')
fig.colorbar(surf, ax=ax1, shrink=0.5, aspect=5)

# 2. Contour plot of D(T, C)
ax2 = fig.add_subplot(222)
contour = ax2.contour(T_mesh, C_mesh, D_mesh, levels=20)
ax2.clabel(contour, inline=True, fontsize=8)
ax2.set_xlabel('Temperature (T)')
ax2.set_ylabel('Cycle Count (C)')
ax2.set_title('Degradation Contour Lines')

# 3. Time evolution of T(t), C(t), and D(t)
t_range = np.linspace(0, 5, 100)
T_values = 2*t_range
C_values = 5*t_range
D_values = 0.01*T_values**2 + 0.05*C_values

ax3 = fig.add_subplot(223)
ax3.plot(t_range, T_values, 'r-', label='T(t) = 2t')
ax3.plot(t_range, C_values, 'b-', label='C(t) = 5t')
ax3.plot(t_range, D_values, 'g-', label='D(t)')
ax3.axvline(x=3, color='k', linestyle='--', alpha=0.7, label='t = 3')
ax3.set_xlabel('Time (t)')
ax3.set_ylabel('Value')
ax3.set_title('Time Evolution')
ax3.legend()
ax3.grid(True)

# 4. Rate of degradation over time
dD_dt_values = []
for t_val in t_range:
    dD_dt_values.append(float(dD_dt_simplified.subs(t, t_val)))
dD_dt_values = np.array(dD_dt_values)

ax4 = fig.add_subplot(224)
ax4.plot(t_range, dD_dt_values, 'purple', linewidth=2)
ax4.axvline(x=3, color='k', linestyle='--', alpha=0.7, label='t = 3')
ax4.axhline(y=float(dD_dt_at_3), color='r', linestyle=':', label=f'dD/dt at t=3 = {float(dD_dt_at_3):.4f}')
ax4.set_xlabel('Time (t)')
ax4.set_ylabel('dD/dt')
ax4.set_title('Rate of Degradation Over Time')
ax4.legend()
ax4.grid(True)

plt.tight_layout()
plt.savefig('battery_degradation_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# Summary
print("\n" + "="*50)
print("SUMMARY")
print("="*50)
print(f"1. The degradation function: D(T,C) = 0.01T² + 0.05C")
print(f"2. Partial derivatives:")
print(f"   ∂D/∂T = {dD_dT}")
print(f"   ∂D/∂C = {dD_dC}")
print(f"3. Time-dependent functions:")
print(f"   T(t) = {T_t}")
print(f"   C(t) = {C_t}")
print(f"4. Total rate of degradation:")
print(f"   dD/dt = {dD_dt_simplified}")
print(f"5. At t = 3:")
print(f"   T(3) = {float(T_t.subs(t, 3))}")
print(f"   C(3) = {float(C_t.subs(t, 3))}")
print(f"   dD/dt|_{{t=3}} = {float(dD_dt_at_3)}")
print(f"   D(3) = {float(D.subs([(T, T_t.subs(t, 3)), (C, C_t.subs(t, 3))]))}")
print("="*50)
```

# Battery Degradation: A Chain Rule and Partial Derivatives Example

I'll walk you through this step-by-step problem involving partial derivatives and the chain rule, demonstrating both the mathematical theory and practical implementation with Python.

Let's start by understanding what we're working with. We have a battery degradation function that depends on both temperature and cycle count. This is a great real-world example of how multivariable calculus applies to engineering problems.

Now let me provide a detailed explanation of each step to help you understand the mathematical concepts behind this problem.

---

## Understanding the Problem

We're analyzing how battery degradation changes over time. The degradation depends on two factors: temperature (T) and cycle count (C). Both of these factors increase with time according to given functions. We want to find out how quickly the battery is degrading at a specific moment in time.

---

## Step 1: The Degradation Function

Our degradation function is:  
**D(T, C) = 0.01T² + 0.05C**

This means:
- Temperature has a quadratic effect (T² term) - small temperature increases cause larger degradation increases
- Cycle count has a linear effect - each additional cycle adds a constant amount of degradation
- The coefficients (0.01 and 0.05) represent the relative impact of each factor

---

## Step 2: Partial Derivatives

When we compute partial derivatives, we're asking:  
**"If I change just one variable while keeping the other constant, how does the function change?"**

- **∂D/∂T = 0.02T**  
  This tells us that for each unit increase in temperature (while keeping cycles constant), degradation increases by 0.02T units. Notice this depends on the current temperature — higher temperatures cause more degradation per degree increase.

- **∂D/∂C = 0.05**  
  This tells us that for each additional cycle (while keeping temperature constant), degradation increases by 0.05 units. This is constant regardless of how many cycles have already occurred.

---

## Step 3: Time-Dependent Functions

In real life, both temperature and cycle count change over time:
- **T(t) = 2t**: Temperature increases by 2 degrees per unit time
- **C(t) = 5t**: Cycle count increases by 5 cycles per unit time

These linear relationships mean both factors grow at constant rates over time.

---

## Step 4: The Chain Rule

Here's where it gets interesting! The chain rule helps us find how D changes with time when D depends on variables (T and C) that themselves change with time.

The formula is:
**dD/dt = (∂D/∂T)(dT/dt) + (∂D/∂C)(dC/dt)**

This makes intuitive sense:
- The first term represents the contribution to degradation rate from temperature changes
- The second term represents the contribution from cycle count changes
- We add them because both are happening simultaneously

### Let's calculate each component:

1. **From temperature changes**:
   - ∂D/∂T = 0.02T = 0.02(2t) = **0.04t**
   - dT/dt = 2  
   - Contribution: **0.04t × 2 = 0.08t**

2. **From cycle count changes**:
   - ∂D/∂C = 0.05  
   - dC/dt = 5  
   - Contribution: **0.05 × 5 = 0.25**

### Total rate:  
**dD/dt = 0.08t + 0.25**

Notice that the degradation rate itself changes with time due to the temperature effect, while the cycle effect remains constant.

---

## Step 5: Evaluation at t = 3

At **t = 3**:
- **Temperature**: T(3) = 2(3) = **6 degrees**
- **Cycle count**: C(3) = 5(3) = **15 cycles**
- **Degradation rate**:  
  dD/dt|_{t=3} = 0.08(3) + 0.25 = **0.24 + 0.25 = 0.49**

This means at t = 3, the battery is degrading at a rate of **0.49 degradation units per time unit**.

---

4. **Degradation Rate**:  
   Shows that dD/dt increases linearly with time. At t = 3, the rate is 0.49 units per time unit.

---

## Physical Interpretation

This model suggests that:
- **Temperature has an accelerating effect on degradation** (due to the T² term)
- **Cycling has a steady, cumulative effect**
- **As time progresses, the degradation rate increases**, meaning the battery degrades faster as it ages

This matches real-world battery behavior where higher temperatures and more charge-discharge cycles both contribute to battery wear, with temperature effects often being more severe.

# Reflection Questions: Understanding Partial Derivatives and the Chain Rule

These are excellent questions that get to the heart of what makes multivariable calculus so powerful for understanding real-world systems. Let me explore each one in depth to help you build a complete mental picture of these concepts.

## 1. How do partial derivatives help isolate the influence of each variable?

Think of partial derivatives as conducting a controlled experiment in mathematics. When you're trying to understand how multiple variables affect an outcome, you face the same challenge a scientist does: how can you figure out what each factor contributes individually when everything is changing at once?

### The Isolation Principle

When we calculate ∂D/∂T (the partial derivative with respect to temperature), we're essentially asking: "What happens if I adjust *only* temperature by a tiny amount, while freezing everything else in place?" This is like a chemist changing one ingredient in a reaction while keeping all others constant.

In our battery example:
- **∂D/∂T = 0.02T** tells us that for each unit increase in temperature (with cycles held constant), degradation increases by 0.02T units
- **∂D/∂C = 0.05** tells us that for each additional cycle (with temperature frozen), degradation increases by 0.05 units

### Why This Matters

Without partial derivatives, we'd only see the combined effect of all variables changing together - like trying to taste individual spices in a complex dish. Partial derivatives let us "untangle" these effects and understand:

1. Which variables have the strongest influence
2. How the influence changes as conditions change
3. Whether variables interact in unexpected ways

For instance, notice that ∂D/∂T = 0.02T means temperature's impact depends on how hot it already is. At T = 100°, a 1-degree increase causes twice as much degradation as when T = 50°. This nonlinear behavior would be impossible to understand without isolating temperature's effect.

### A Mental Exercise

Imagine you're optimizing battery life. Partial derivatives give you a "sensitivity map" - they show you exactly where small adjustments will have the biggest impact. Should you focus on cooling systems (reducing T) or usage patterns (reducing C)? The partial derivatives give you precise guidance.

## 2. What does the chain rule reveal about indirect changes?

The chain rule is like discovering that your actions create ripple effects. It's one thing to understand direct relationships (like how pushing a door directly affects its motion), but the chain rule helps us understand cascading effects (like how pushing one domino affects the entire chain).

### Understanding Indirect Influence

In our battery example, neither temperature nor cycle count directly depends on time - but both change *because* time passes. The chain rule lets us trace these indirect connections:

**dD/dt = (∂D/∂T)(dT/dt) + (∂D/∂C)(dC/dt)**

This formula is profound because it reveals:

1. **Decomposition of effects**: Degradation changes with time through two pathways - via temperature and via cycles
2. **Multiplication of rates**: The effect isn't just additive; it's the product of sensitivity (∂D/∂T) and rate of change (dT/dt)
3. **Hidden dependencies**: Time doesn't appear in our original function D(T,C), yet it influences degradation through its effect on T and C

### The Cascade Metaphor

Think of it like a waterfall with multiple streams:
- Time flows into both temperature and cycles
- These intermediate variables then flow into degradation
- The chain rule tells us how to calculate the total flow

### Real-World Implications

The chain rule reveals important insights about systems:

1. **Leverage points**: If you want to reduce dD/dt, you can either decrease the sensitivities (∂D/∂T or ∂D/∂C) or slow down the rates (dT/dt or dC/dt)
2. **Amplification effects**: Small changes in T or C can have outsized impacts if the partial derivatives are large
3. **System dynamics**: The chain rule shows how a system evolves through intermediate variables, revealing potential intervention points

### A Thinking Question

Consider this: What if temperature and cycles weren't independent? What if higher temperatures caused more frequent cycling? The chain rule would reveal even more complex interdependencies.

## 3. How would dD/dt change if temperature increased faster than cycle count?

This question beautifully illustrates how the relative rates of change affect overall system behavior. Let's explore different scenarios to build intuition.

### Current Scenario Baseline

Currently we have:
- T(t) = 2t → dT/dt = 2
- C(t) = 5t → dC/dt = 5
- dD/dt = 0.08t + 0.25

Notice that cycles are increasing faster than temperature (5 vs 2 units per time).

### Scenario: Faster Temperature Increase

Let's say temperature now increases faster: T(t) = 4t while cycles remain C(t) = 5t

**New calculations:**
- dT/dt = 4
- dC/dt = 5 (unchanged)
- ∂D/∂T = 0.02T = 0.02(4t) = 0.08t
- ∂D/∂C = 0.05 (unchanged)

**New degradation rate:**
dD/dt = (∂D/∂T)(dT/dt) + (∂D/∂C)(dC/dt)
dD/dt = (0.08t)(4) + (0.05)(5)
dD/dt = 0.32t + 0.25

### Comparing the Impact

| Time | Original dD/dt | New dD/dt | Difference |
|------|----------------|-----------|------------|
| t=1  | 0.33          | 0.57      | +0.24      |
| t=3  | 0.49          | 1.21      | +0.72      |
| t=5  | 0.65          | 1.85      | +1.20      |

### Key Insights

1. **Nonlinear amplification**: Notice how the difference grows over time. This happens because temperature has a quadratic effect (T²), so faster temperature increase creates an accelerating impact.

2. **Dominant factor shifting**: In the original scenario, at early times (t≈0), cycles dominated (0.25 vs 0). With faster temperature increase, temperature quickly becomes dominant.

3. **Compound effects**: The change in degradation rate isn't just proportional to the change in temperature rate. The formula dD/dt = 0.32t + 0.25 shows the coefficient of t has quadrupled (from 0.08 to 0.32), creating an even more dramatic acceleration.

### Physical Interpretation

If we implemented better thermal insulation, temperature might rise more slowly. Conversely, if cooling systems fail, temperature could rise faster than cycling rate. The mathematics reveals that:

- **Temperature acceleration has outsized impact** due to the quadratic relationship
- **Early intervention matters more** when temperature rises quickly
- **The system becomes increasingly unstable** as time progresses with faster temperature increase

### A Design Insight

This analysis suggests that for long-lived batteries, thermal management becomes increasingly critical over time. A small improvement in cooling that reduces dT/dt could have dramatic effects on battery lifespan, especially compared to reducing cycling frequency.
